```python
jobs = {
    "A": 3,
    "B": 5,
    "C": 2,
    "D": 4
}


def calculate_cost(schedule):
    total = 0
    completion_time = 0

    for job in schedule:
        completion_time += jobs[job]
        total += completion_time

    return total


def get_neighbors(schedule):
    neighbors = []

    for i in range(len(schedule)):
        for j in range(i + 1, len(schedule)):
            neighbor = schedule.copy()

            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

            neighbors.append(neighbor)

    return neighbors


current = list(jobs.keys())

print("Initial Schedule:", current)
print("Initial Cost:", calculate_cost(current))


while True:

    current_cost = calculate_cost(current)

    neighbors = get_neighbors(current)

    best_neighbor = current
    best_cost = current_cost

    for neighbor in neighbors:

        neighbor_cost = calculate_cost(neighbor)

        if neighbor_cost < best_cost:
            best_neighbor = neighbor
            best_cost = neighbor_cost

    if best_cost < current_cost:
        current = best_neighbor
        print("Better schedule found:", current)
        print("Cost:", best_cost)
    else:
        break


print("\nFINAL SCHEDULE")
print("Schedule:", current)
print("Cost:", calculate_cost(current))
```
