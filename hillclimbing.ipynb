```python
# Hill Climbing for Maximizing a Mathematical Function

def f(x):
    return -(x - 4) ** 2 + 16


# Initial state
current = 0

print("Function: f(x) = -(x - 4)^2 + 16")
print("Initial State:", current)
print("Initial Value:", f(current))

while True:

    # Generate neighbours
    left = current - 1
    right = current + 1

    # Find the best neighbour
    if f(left) > f(current):
        next_state = left
    elif f(right) > f(current):
        next_state = right
    else:
        next_state = current

    print("\nCurrent State:", current)
    print("Current Value:", f(current))
    print("Left Neighbour:", left, "Value:", f(left))
    print("Right Neighbour:", right, "Value:", f(right))

    # If no better neighbour exists, stop
    if next_state == current:
        break

    current = next_state

print("\nFinal Best State:", current)
print("Maximum Value:", f(current))
```
