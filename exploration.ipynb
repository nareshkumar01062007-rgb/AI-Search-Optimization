```python
import random
solutions = list(range(1, 11))

current = random.choice(solutions)

print("Initial Solution:", current)

for step in range(10):
    explored = random.choice(solutions)
    exploited = min(current + 1, 10)

    print("\nStep", step + 1)
    print("Exploration:", explored)
    print("Exploitation:", exploited)

    if explored > exploited:
        current = explored
    else:
        current = exploited

    print("Current Best:", current)

print("\nFinal Best Solution:", current)
```
