```python
import math
import random


def f(x):
    return -(x - 4) ** 2 + 16


# Initial state
current = 0

# Initial temperature
temperature = 10

# Cooling rate
cooling_rate = 0.5

print("Function: f(x) = -(x - 4)^2 + 16")
print("Initial State:", current)
print("Initial Value:", f(current))

while temperature > 0.1:

    # Generate a random neighbour
    move = random.choice([-1, 1])
    next_state = current + move

    current_value = f(current)
    next_value = f(next_state)

    # Difference in value
    delta = next_value - current_value

    # Accept better solution
    if delta > 0:
        current = next_state
        decision = "Better state accepted"

    # Sometimes accept worse solution
    else:
        probability = math.exp(delta / temperature)
        random_value = random.random()

        if random_value < probability:
            current = next_state
            decision = "Worse state accepted"
        else:
            decision = "State rejected"

    print(
        "\nTemperature:", round(temperature, 2),
        " Current:", current,
        " Value:", f(current),
        " ->", decision
    )

    # Reduce temperature
    temperature = temperature * cooling_rate


print("\nFinal Best State:", current)
print("Maximum Value:", f(current))
```
