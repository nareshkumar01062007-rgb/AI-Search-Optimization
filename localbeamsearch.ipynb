```python
# Local Beam Search for Maximizing a Mathematical Function

def f(x):
    return -(x - 4) ** 2 + 16


k = 3

# Initial states
states = [0, 2, 6]

print("Function: f(x) = -(x - 4)^2 + 16")
print("Initial States:", states)

while True:

    # Generate neighbours
    candidates = []

    for x in states:
        candidates.append(x - 1)
        candidates.append(x + 1)

    # Remove duplicates
    candidates = list(set(candidates))

    # Sort according to function value
    candidates.sort(key=f, reverse=True)

    # Select best k states
    new_states = candidates[:k]

    print("\nCandidates:", candidates)
    print("Selected States:", new_states)

    # Check whether optimum is reached
    if max(f(x) for x in new_states) == 16:
        states = new_states
        break

    states = new_states

best = max(states, key=f)

print("\nFinal Best State:", best)
print("Maximum Value:", f(best))
```
